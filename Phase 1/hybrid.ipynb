```python
class ByteLevelBPETokenizer:
    SPECIAL_TOKENS = ["<pad>", "<unk>", "<s>", "</s>"]

    def __init__(self):
        self.merges = {}
        self.id_to_bytes = {i: bytes([i]) for i in range(256)}
        self.special_to_id = {}
        for i, tok in enumerate(self.SPECIAL_TOKENS):
            tid = 256 + i
            self.special_to_id[tok] = tid
            self.id_to_bytes[tid] = tok.encode("utf-8")

    @property
    def vocab_size(self):
        return len(self.id_to_bytes)

    @staticmethod
    def _pair_counts(sequences):
        counts = {}
        for seq, freq in sequences:
            for pair in zip(seq, seq[1:]):
                counts[pair] = counts.get(pair, 0) + freq
        return counts

    @staticmethod
    def _apply_merge(seq, pair, new_id):
        out, i = [], 0
        while i < len(seq):
            if i < len(seq) - 1 and (seq[i], seq[i + 1]) == pair:
                out.append(new_id); i += 2
            else:
                out.append(seq[i]); i += 1
        return out

    def train(self, text, vocab_size, max_words=MAX_BPE_WORDS, verbose=True):
        num_merges = vocab_size - self.vocab_size
        if num_merges <= 0:
            return self

        word_freq = {}
        for word in text.split():
            word_freq[word] = word_freq.get(word, 0) + 1
        if len(word_freq) > max_words:
            word_freq = dict(sorted(word_freq.items(), key=lambda kv: -kv[1])[:max_words])
        sequences = [[list(word.encode("utf-8")), freq] for word, freq in word_freq.items()]

        next_id = self.vocab_size
        for step in range(num_merges):
            counts = self._pair_counts(sequences)
            if not counts:
                break
            best_pair = max(counts, key=counts.get)
            self.merges[best_pair] = next_id
            self.id_to_bytes[next_id] = self.id_to_bytes[best_pair[0]] + self.id_to_bytes[best_pair[1]]
            for entry in sequences:
                entry[0] = self._apply_merge(entry[0], best_pair, next_id)
            if verbose and (step + 1) % 1000 == 0:
                print(step + 1, next_id + 1)
            next_id += 1
        return self

    def encode(self, text):
        ids = []
        for word in text.split():
            seq = list(word.encode("utf-8"))
            while len(seq) > 1:
                mergeable = [p for p in zip(seq, seq[1:]) if p in self.merges]
                if not mergeable:
                    break
                pair = min(mergeable, key=lambda p: self.merges[p])
                seq = self._apply_merge(seq, pair, self.merges[pair])
            ids.extend(seq)
            ids.append(self.special_to_id["</s>"])
        return ids

    def encode_to_array(self, text, max_tokens):
        arr = np.empty(max_tokens, dtype=np.int32)
        n = 0
        for word in text.split():
            seq = list(word.encode("utf-8"))
            while len(seq) > 1:
                mergeable = [p for p in zip(seq, seq[1:]) if p in self.merges]
                if not mergeable:
                    break
                pair = min(mergeable, key=lambda p: self.merges[p])
                seq = self._apply_merge(seq, pair, self.merges[pair])
            seq.append(self.special_to_id["</s>"])
            for t in seq:
                if n >= max_tokens:
                    return arr[:n]
                arr[n] = t; n += 1
        return arr[:n]

    def decode(self, ids):
        words, current = [], []
        for i in ids:
            if i == self.special_to_id["</s>"]:
                if current:
                    words.append(b"".join(self.id_to_bytes.get(int(t), b"") for t in current).decode("utf-8", errors="ignore"))
                    current = []
            else:
                current.append(i)
        if current:
            words.append(b"".join(self.id_to_bytes.get(int(t), b"") for t in current).decode("utf-8", errors="ignore"))
        return " ".join(words)

    def save(self, path):
        merges_serializable = {f"{a},{b}": nid for (a, b), nid in self.merges.items()}
        with open(path, "w", encoding="utf-8") as f:
            json.dump({"merges": merges_serializable}, f)

    @classmethod
    def load(cls, path):
        tok = cls()
        with open(path, "r", encoding="utf-8") as f:
            data = json.load(f)
        for pair_str, nid in data["merges"].items():
            a, b = (int(x) for x in pair_str.split(","))
            tok.merges[(a, b)] = nid
            tok.id_to_bytes[nid] = tok.id_to_bytes[a] + tok.id_to_bytes[b]
        return tok
```

```python
class RMSNorm(nn.Module):
    """RMS normalization without mean subtraction."""
    def __init__(self, dim, eps=1e-6):
        super().__init__()
        self.weight = nn.Parameter(torch.ones(dim))
        self.eps = eps

    def forward(self, x):
        rms = x.pow(2).mean(dim=-1, keepdim=True)
        return self.weight * x * torch.rsqrt(rms + self.eps)


def rotate_half(x):
    x1, x2 = x.chunk(2, dim=-1)
    return torch.cat((-x2, x1), dim=-1)


class RotaryEmbedding(nn.Module):
    def __init__(self, dim, max_seq_len, base=10000.0):
        super().__init__()
        inv_freq = 1.0 / (base ** (torch.arange(0, dim, 2, dtype=torch.float32) / dim))
        freqs = torch.outer(torch.arange(max_seq_len, dtype=torch.float32), inv_freq)
        emb = torch.cat((freqs, freqs), dim=-1)
        self.register_buffer("cos", emb.cos()[None, None, :, :], persistent=False)
        self.register_buffer("sin", emb.sin()[None, None, :, :], persistent=False)

    def forward(self, x, offset=0):
        T = x.size(-2)
        cos = self.cos[:, :, offset:offset + T, :].to(x.dtype)
        sin = self.sin[:, :, offset:offset + T, :].to(x.dtype)
        return x * cos + rotate_half(x) * sin


class GeGLU(nn.Module):
    def __init__(self, d_model, d_ff, dropout):
        super().__init__()
        self.gate = nn.Linear(d_model, d_ff, bias=False)
        self.up = nn.Linear(d_model, d_ff, bias=False)
        self.down = nn.Linear(d_ff, d_model, bias=False)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        return self.dropout(self.down(F.gelu(self.gate(x)) * self.up(x)))


class GatedCausalConvMixer(nn.Module):
    """
    Local token mixer (Conformer / ConvBERT / gated-conv style) made causal for LM:
        x -> [value | gate] projection -> causal depthwise Conv1d(k) on value -> value * silu(gate) -> out proj
    Receptive field per layer = kernel size; no positional embedding needed (convolution is position-aware).
    Cost is O(T * k) instead of O(T^2).
    """
    def __init__(self, d_model, kernel_size, dropout):
        super().__init__()
        self.k = kernel_size
        self.in_proj = nn.Linear(d_model, 2 * d_model, bias=False)
        self.dw_weight = nn.Parameter(torch.empty(d_model, 1, kernel_size))
        nn.init.normal_(self.dw_weight, std=0.02)
        self.out_proj = nn.Linear(d_model, d_model, bias=False)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x, cache=None, offset=0):
        B, T, C = x.shape
        a, g = self.in_proj(x).chunk(2, dim=-1)
        a = a.transpose(1, 2)
        if cache is None:
            state = a.new_zeros(B, C, self.k - 1)
        else:
            state = cache["state"]
        a_pad = torch.cat([state, a], dim=2)
        y = F.conv1d(a_pad, self.dw_weight.to(a.dtype), groups=C)
        new_cache = {"state": a_pad[:, :, a_pad.size(2) - (self.k - 1):]}
        y = y.transpose(1, 2) * F.silu(g)
        return self.dropout(self.out_proj(y)), new_cache


class CausalSelfAttention(nn.Module):
    """Global causal multi-head attention with RoPE; used sparsely (every ATTN_EVERY-th layer)."""
    def __init__(self, d_model, n_heads, rope, dropout):
        super().__init__()
        assert d_model % n_heads == 0
        self.n_heads, self.d_head = n_heads, d_model // n_heads
        self.q_proj = nn.Linear(d_model, d_model, bias=False)
        self.k_proj = nn.Linear(d_model, d_model, bias=False)
        self.v_proj = nn.Linear(d_model, d_model, bias=False)
        self.out_proj = nn.Linear(d_model, d_model, bias=False)
        self.rope, self.dropout = rope, dropout
        self.resid_dropout = nn.Dropout(dropout)

    def _split(self, x):
        B, T, _ = x.shape
        return x.view(B, T, self.n_heads, self.d_head).transpose(1, 2)

    def forward(self, x, cache=None, offset=0):
        B, T, C = x.shape
        q, k, v = self._split(self.q_proj(x)), self._split(self.k_proj(x)), self._split(self.v_proj(x))
        q, k = self.rope(q, offset), self.rope(k, offset)
        if cache is None:
            out = F.scaled_dot_product_attention(q, k, v, dropout_p=self.dropout if self.training else 0.0, is_causal=True)
        else:
            assert T == 1
            k, v = torch.cat([cache["k"], k], dim=2), torch.cat([cache["v"], v], dim=2)
            out = F.scaled_dot_product_attention(q, k, v, is_causal=False)
        out = out.transpose(1, 2).contiguous().view(B, T, C)
        return self.resid_dropout(self.out_proj(out)), {"k": k, "v": v}


class HybridBlock(nn.Module):
    def __init__(self, d_model, n_heads, d_ff, rope, dropout, use_attention, conv_kernel):
        super().__init__()
        self.use_attention = use_attention
        self.norm_mix = RMSNorm(d_model)
        self.mixer = CausalSelfAttention(d_model, n_heads, rope, dropout) if use_attention \
            else GatedCausalConvMixer(d_model, conv_kernel, dropout)
        self.norm_ffn = RMSNorm(d_model)
        self.ffn = GeGLU(d_model, d_ff, dropout)

    def forward(self, x, cache=None, offset=0):
        h, new_cache = self.mixer(self.norm_mix(x), cache=cache, offset=offset)
        x = x + h
        x = x + self.ffn(self.norm_ffn(x))
        return x, new_cache


class HybridConvAttnLM(nn.Module):
    """
    Encoder-style (conformer/ConvBERT-like) hybrid adapted to causal language modelling:
    layer pattern for attn_every=4, 12 layers:  C C C A | C C C A | C C C A
    (C = gated causal depthwise-conv mixer, A = global causal RoPE attention), GeGLU FFN everywhere.
    """
    def __init__(self, vocab_size, d_model, n_layers, n_heads, d_ff, max_seq_len,
                 attn_every=4, conv_kernel=15, dropout=0.1, rope_base=10000.0, tie_weights=True):
        super().__init__()
        self.max_seq_len = max_seq_len
        self.token_emb = nn.Embedding(vocab_size, d_model)
        self.emb_dropout = nn.Dropout(dropout)
        self.rope = RotaryEmbedding(d_model // n_heads, max_seq_len, rope_base)
        self.blocks = nn.ModuleList([
            HybridBlock(d_model, n_heads, d_ff, self.rope, dropout,
                        use_attention=((i + 1) % attn_every == 0), conv_kernel=conv_kernel)
            for i in range(n_layers)
        ])
        self.final_norm = RMSNorm(d_model)
        self.lm_head = nn.Linear(d_model, vocab_size, bias=False)
        if tie_weights:
            self.lm_head.weight = self.token_emb.weight
        self.apply(self._init_weights)

    def _init_weights(self, module):
        if isinstance(module, (nn.Linear, nn.Embedding)):
            nn.init.normal_(module.weight, mean=0.0, std=0.02)

    def _run(self, idx, caches=None, offset=0):
        B, T = idx.shape
        if offset + T > self.max_seq_len:
            raise ValueError(f"Sequence length {offset + T} exceeds max_seq_len={self.max_seq_len}")
        x = self.emb_dropout(self.token_emb(idx))
        new_caches = []
        for i, blk in enumerate(self.blocks):
            x, c = blk(x, cache=None if caches is None else caches[i], offset=offset)
            new_caches.append(c)
        return self.lm_head(self.final_norm(x)), new_caches

    def forward(self, idx, targets=None):
        logits, _ = self._run(idx)
        loss = None
        if targets is not None:
            loss = F.cross_entropy(logits.reshape(-1, logits.size(-1)), targets.reshape(-1))
        return logits, loss

    @torch.no_grad()
    def generate(self, idx, max_new_tokens, temperature=1.0, top_p=0.95, eos_id=None, stop_after_eos=None):
        """Prefill the prompt once, then decode token-by-token using conv states + attention KV caches.
        Returns ONLY the newly generated tokens. Batch size 1 for early stopping."""
        self.eval()
        assert idx.size(1) + max_new_tokens <= self.max_seq_len
        logits, caches = self._run(idx)
        P, out, n_eos = idx.size(1), [], 0
        for t in range(max_new_tokens):
            probs = torch.softmax(logits[:, -1, :] / max(temperature, 1e-5), dim=-1)
            sorted_probs, sorted_idx = torch.sort(probs, descending=True, dim=-1)
            cumulative = torch.cumsum(sorted_probs, dim=-1)
            remove = cumulative > top_p
            remove[:, 1:] = remove[:, :-1].clone()
            remove[:, 0] = False
            sorted_probs = sorted_probs.masked_fill(remove, 0.0)
            sorted_probs = sorted_probs / sorted_probs.sum(dim=-1, keepdim=True)
            nxt = torch.gather(sorted_idx, -1, torch.multinomial(sorted_probs, 1))
            out.append(nxt)
            if stop_after_eos is not None and eos_id is not None and idx.size(0) == 1:
                n_eos += int(nxt.item() == eos_id)
                if n_eos >= stop_after_eos:
                    break
            if t < max_new_tokens - 1:
                logits, caches = self._run(nxt, caches=caches, offset=P + t)
        return torch.cat(out, dim=1)
```

```python
def build_model(arch_name):
    if arch_name == "hybrid_conv_attn":
        return HybridConvAttnLM(
            vocab_size=VOCAB_SIZE, d_model=D_MODEL, n_layers=N_LAYERS, n_heads=N_HEADS, d_ff=D_FF,
            max_seq_len=BLOCK_SIZE, attn_every=ATTN_EVERY, conv_kernel=CONV_KERNEL,
            dropout=DROPOUT, rope_base=ROPE_BASE, tie_weights=True,
        )
    raise ValueError(f"Unknown architecture: {arch_name}")


raw_model = build_model(ARCH_NAME).to(DEVICE)

n_params = sum(p.numel() for p in raw_model.parameters())
n_attn = sum(b.use_attention for b in raw_model.blocks)
print(f"Parameters: {n_params / 1e6:.2f}M | attention layers: {n_attn}/{N_LAYERS} | conv layers: {N_LAYERS - n_attn}/{N_LAYERS}")
assert 50e6 <= n_params <= 300e6

if NUM_GPUS > 1:
    model = nn.DataParallel(raw_model)
else:
    model = raw_model
```

### Architecture summary
- **Model:** HybridConvAttnLM
- **Layers:** 12; 9 gated causal depthwise-convolution layers and 3 global causal RoPE-attention layers
- **Pattern:** C C C A, repeated three times
- **Normalization / FFN:** RMSNorm and GeGLU
- **Context:** 1024 tokens
- **Generation:** convolution states and attention KV cache

```python
def get_lr(step, total_steps, warmup_steps, base_lr):
    if step < warmup_steps:
        return base_lr * step / max(1, warmup_steps)
    progress = (step - warmup_steps) / max(1, total_steps - warmup_steps)
    return 0.5 * base_lr * (1 + math.cos(math.pi * progress))

def cycle_loader(loader):
    while True:
        for batch in loader:
            yield batch


def evaluate(loader, max_batches=50):
    model.eval()
    losses = []
    with torch.no_grad():
        for i, (x, y) in enumerate(loader):
            if i >= max_batches:
                break
            x, y = x.to(DEVICE), y.to(DEVICE)
            with torch.autocast(device_type="cuda", enabled=torch.cuda.is_available()):
                _, loss = model(x, y)
                loss = loss.mean()
            losses.append(loss.item())
    model.train()
    return sum(losses) / max(1, len(losses))

def save_checkpoint(path, step, train_loss_history, val_loss_history, step_history, elapsed_seconds):
    tmp_path = path + ".tmp"
    torch.save({
        "step": step, "model_state": raw_model.state_dict(), "optimizer_state": optimizer.state_dict(),
        "scaler_state": scaler.state_dict(), "train_loss_history": train_loss_history,
        "val_loss_history": val_loss_history, "step_history": step_history, "elapsed_seconds": elapsed_seconds,
        "torch_rng_state": torch.get_rng_state(),
        "cuda_rng_state": torch.cuda.get_rng_state_all() if torch.cuda.is_available() else None,
        "python_rng_state": random.getstate(),
        "run_config": {"ARCH_NAME": ARCH_NAME, "D_MODEL": D_MODEL, "N_LAYERS": N_LAYERS, "N_HEADS": N_HEADS,
                       "D_FF": D_FF, "BLOCK_SIZE": BLOCK_SIZE, "VOCAB_SIZE": VOCAB_SIZE, "ATTN_EVERY": ATTN_EVERY, "CONV_KERNEL": CONV_KERNEL, "ROPE_BASE": ROPE_BASE, "ARCH_NAME": ARCH_NAME},
    }, tmp_path)
    os.replace(tmp_path, path)

def load_checkpoint(path):
    ckpt = torch.load(path, map_location="cpu", weights_only=False)
    cfg = ckpt["run_config"]
    mismatches = [k for k, v in cfg.items() if str(v) != str(globals()[k])]
    if mismatches:
        raise ValueError(mismatches)
    raw_model.load_state_dict(ckpt["model_state"])
    optimizer.load_state_dict(ckpt["optimizer_state"])
    scaler.load_state_dict(ckpt["scaler_state"])
    torch.set_rng_state(ckpt["torch_rng_state"])
    if ckpt["cuda_rng_state"] is not None and torch.cuda.is_available():
        torch.cuda.set_rng_state_all(ckpt["cuda_rng_state"])
    random.setstate(ckpt["python_rng_state"])
    return ckpt["step"], ckpt["train_loss_history"], ckpt["val_loss_history"], ckpt["step_history"], ckpt["elapsed_seconds"]


optimizer = torch.optim.AdamW(raw_model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY, betas=(0.9, 0.95))
scaler = torch.amp.GradScaler("cuda", enabled=torch.cuda.is_available())

if os.path.exists(CHECKPOINT_PATH):
    step, train_loss_history, val_loss_history, step_history, prev_elapsed = load_checkpoint(CHECKPOINT_PATH)
else:
    step, train_loss_history, val_loss_history, step_history, prev_elapsed = 0, [], [], [], 0.0

t0 = time.time()
model.train()
data_iter = iter(cycle_loader(train_loader))

while step < TRAIN_STEPS:
    optimizer.zero_grad(set_to_none=True)
    accum_loss = 0.0
    for _ in range(GRAD_ACCUM_STEPS):
        x, y = next(data_iter)
        x, y = x.to(DEVICE), y.to(DEVICE)
        with torch.autocast(device_type="cuda", enabled=torch.cuda.is_available()):
            _, loss = model(x, y)
            loss = loss.mean() / GRAD_ACCUM_STEPS
        scaler.scale(loss).backward()
        accum_loss += loss.item()

    scaler.unscale_(optimizer)
    torch.nn.utils.clip_grad_norm_(raw_model.parameters(), 1.0)
    lr_now = get_lr(step, TRAIN_STEPS, WARMUP_STEPS, LR)
    for g in optimizer.param_groups:
        g["lr"] = lr_now
    scaler.step(optimizer)
    scaler.update()
    step += 1

    if step % 50 == 0:
        train_loss_history.append(accum_loss); step_history.append(step)
        print(step, accum_loss, lr_now)

    if step % EVAL_EVERY == 0 or step == TRAIN_STEPS:
        v_loss = evaluate(val_loader)
        val_loss_history.append((step, v_loss))
        print("eval", step, v_loss, math.exp(v_loss))

    if step % CHECKPOINT_EVERY == 0 or step == TRAIN_STEPS:
        elapsed = prev_elapsed + (time.time() - t0)
        save_checkpoint(CHECKPOINT_PATH, step, train_loss_history, val_loss_history, step_history, elapsed)

training_time = prev_elapsed + (time.time() - t0)
print(training_time / 3600)
```

```python
def trim_to_words(text, n_words):
    return " ".join(text.split()[:n_words])

EOS_ID = tokenizer.special_to_id["</s>"]

def generate_continuation(prompt_text, max_new_tokens=MAX_GENERATION_TOKENS):
    prompt_ids = tokenizer.encode(prompt_text)[-(BLOCK_SIZE - max_new_tokens):]
    idx = torch.tensor([prompt_ids], dtype=torch.long, device=DEVICE)
    out = raw_model.generate(idx, max_new_tokens=max_new_tokens, temperature=TEMPERATURE, top_p=TOP_P,
                             eos_id=EOS_ID, stop_after_eos=TARGET_GENERATION_WORDS + 5)
    new_text = tokenizer.decode(out[0].tolist())
    return trim_to_words(new_text, TARGET_GENERATION_WORDS)

def generate_for_dataset(prompt_dict):
    predictions = {}
    for i, (ex_id, rec) in enumerate(prompt_dict.items()):
        predictions[ex_id] = {"text": generate_continuation(rec["text"])}
        if (i + 1) % 25 == 0:
            print(i + 1, len(prompt_dict))
    return predictions

raw_model.eval()
```
